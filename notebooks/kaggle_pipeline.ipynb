# Kaggle pipeline for industrial surface defect detection

This notebook is a template for running the dataset validation, audit, baseline benchmarking, and reporting workflow on Kaggle.


```python
!pip install -q -r ../requirements.txt
!pip install -q -e ..

from pathlib import Path
import subprocess
import sys

REPO_ROOT = Path('/kaggle/working/surface-defect-detection')
DATA_ROOT = Path('/kaggle/input/combined-surface-defect-yolo')  # Change this folder name if needed.
PYTHON = sys.executable

def run(*args):
    command = [PYTHON, *map(str, args)]
    print('>', ' '.join(command))
    subprocess.run(command, cwd=REPO_ROOT, check=True)

run(REPO_ROOT / 'scripts/validate_dataset.py', '--data-root', DATA_ROOT, '--json-output', REPO_ROOT / 'reports/validation.json')
run(REPO_ROOT / 'scripts/audit_dataset.py', '--data-root', DATA_ROOT, '--output-dir', REPO_ROOT / 'reports', '--json-output', REPO_ROOT / 'reports/audit_summary.json')
for model_name in ['yolo11', 'rtdetr', 'yolo-world', 'dino', 'grounding-dino']:
    run(REPO_ROOT / 'scripts/train.py', '--data-root', DATA_ROOT, '--config', REPO_ROOT / 'configs/kaggle.yaml', '--model', model_name, '--experiment', model_name, '--device', '0')
print('Five baseline training runs completed.')
```
